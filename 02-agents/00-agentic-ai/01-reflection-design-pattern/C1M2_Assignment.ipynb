## Exercise 1: `generate_draft` Function

**Objective**:
Write a function called `generate_draft` that takes in a string topic and uses a language model to generate a complete draft essay.

**Inputs**:

* `topic` (str): The essay topic.
* `model` (str, optional): The model identifier to use. Defaults to `"openai:gpt-4o"`.

**Output**:

* A string representing the full draft of the essay.

The setup for calling the LLM using the aisuite library is already provided. Focus on crafting the prompt content. You can reference this setup in later exercises to understand how to interact with the library effectively.


```python
# GRADED FUNCTION: generate_draft

def generate_draft(topic: str, model: str = "openai:gpt-4o") -> str: 
    
    ### START CODE HERE ###

    # Define your prompt here. A multi-line f-string is typically used for this.
    prompt = f"""You are a helpful and skilled academic writer.

        Your task is to generate a complete draft essay on the given topic using a reflection-based process.

        Topic:
        "{topic}"

        Follow these steps carefully:

        Step 1 – Initial Draft:
        Write a full essay draft on the topic. Include:
        - A clear introduction with a thesis statement
        - Well-structured body paragraphs with explanations and examples
        - A concise conclusion that reinforces the main ideas

        Do not mention these steps in the output.

        Step 2 – Self-Reflection and Critique:
        Review your initial draft critically. Identify:
        - Weak or unclear arguments
        - Gaps in logic or structure
        - Redundancy, vagueness, or lack of depth
        - Areas where clarity, flow, or coherence can be improved

        Summarize this critique briefly in your own reasoning, but do not include it in the final answer.

        Step 3 – Revised Final Draft:
        Rewrite the essay based on your critique. The revised draft should:
        - Be clearer, more coherent, and better organized
        - Strengthen arguments and explanations
        - Flow smoothly between paragraphs
        - Use precise and formal academic language

        Output only the final revised essay. Do not include reflections, critiques, or step labels.
        """ 

    ### END CODE HERE ###
    
    # Get a response from the LLM by creating a chat with the client.
    response = CLIENT.chat.completions.create(
        model=model,
        messages=[{"role": "user", "content": prompt}],
        temperature=1.0,
    )

    return response.choices[0].message.content
```

Run the following cell to check your code is working correctly:

```python
# Test your code!
unittests.test_generate_draft(generate_draft)
```

## Exercise 2: `reflect_on_draft` Function

**Objective**:
Write a function called `reflect_on_draft` that takes a previously generated essay draft and uses a language model to provide constructive feedback.

**Inputs**:

* `draft` (str): The essay text to reflect on.
* `model` (str, optional): The model identifier to use. Defaults to `"openai:o4-mini"`.

**Output**:

* A string with feedback in paragraph form.

**Requirements**:

* The feedback should be critical but constructive.
* It should address issues such as structure, clarity, strength of argument, and writing style.
* The function should send the draft to the model and return its response.

You do **not** need to rewrite the essay at this step—just analyze and reflect on it.


```python
# GRADED FUNCTION: reflect_on_draft

def reflect_on_draft(draft: str, model: str = "openai:o4-mini") -> str:

    ### START CODE HERE ###

    # Define your prompt here. A multi-line f-string is typically used for this.
    prompt = f"""
        You are an experienced writing instructor and editor.

        Your task is to critically reflect on the following essay draft and provide constructive feedback.

        Essay Draft:
        "{draft}"

        Analyze the essay holistically and write feedback in clear, well-structured paragraphs. Your feedback should address:

        - Overall structure and organization (introduction, body, conclusion)
        - Clarity and coherence of ideas
        - Strength, depth, and consistency of the argument
        - Use of evidence, examples, or explanations (where applicable)
        - Writing style, tone, and academic clarity
        - Any recurring issues or notable strengths

        Be honest and analytical, but supportive and constructive. Point out areas for improvement and explain why they matter, while also acknowledging what the essay does well.

        Do not rewrite the essay.
        Do not include bullet points or headings.
        Do not summarize the essay.
        Output only the feedback in paragraph form."""

    ### END CODE HERE ###

    # Get a response from the LLM by creating a chat with the client.
    response = CLIENT.chat.completions.create(
        model=model,
        messages=[{"role": "user", "content": prompt}],
        temperature=1.0,
    )

    return response.choices[0].message.content
```

## Exercise 3: `revise_draft` Function

**Objective**:
Implement a function called `revise_draft` that improves a given essay draft based on feedback from a reflection step.

**Inputs**:

* `original_draft` (str): The initial version of the essay.
* `reflection` (str): Constructive feedback or critique on the draft.
* `model` (str, optional): The model identifier to use. Defaults to `"openai:gpt-4o"`.

**Output**:

* A string containing the revised and improved essay.

**Requirements**:

* The revised draft should address the issues mentioned in the feedback.
* It should improve clarity, coherence, argument strength, and overall flow.
* The function should use the feedback to guide the revision, and return only the final revised essay.

In this final exercise, you'll also need to manage the call to the LLM using the CLIENT, as you've practiced in previous exercises.

```python
# GRADED FUNCTION: revise_draft

def revise_draft(original_draft: str, reflection: str, model: str = "openai:gpt-4o") -> str:

    ### START CODE HERE ###

    # Define your prompt here. A multi-line f-string is typically used for this.
    prompt = f""" 
        You are a skilled academic editor and revision specialist.

        Your task is to revise an essay draft using constructive feedback provided from a reflection step.

        Original Essay Draft:
        "{original_draft}"

        Reflection Feedback:
        "{reflection}"

        Carefully review both the original draft and the feedback. Revise the essay so that it clearly addresses the issues raised in the feedback. Your revision should:

        - Improve clarity, coherence, and logical flow
        - Strengthen the thesis and supporting arguments
        - Resolve structural or organizational weaknesses
        - Clarify vague or underdeveloped points
        - Improve transitions and overall readability
        - Refine writing style, tone, and academic precision where needed

        Preserve the original topic and intent of the essay, but do not hesitate to reorganize, rewrite, or expand sections if necessary to meaningfully improve quality.

        Do not explain the changes.
        Do not mention the feedback explicitly.
        Do not include commentary, annotations, or step labels.

        Output only the final revised version of the essay."""

    # Get a response from the LLM by creating a chat with the client.
    response = CLIENT.chat.completions.create(
        model=model,
        messages=[{"role": "user", "content": prompt}],
        temperature=1.0,
    )

    ### END CODE HERE ###

    return response.choices[0].message.content
```
